```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies.head()
```

```python
credits.head()
```

```python
# merging both the data frames as we have two different data frames
movies = movies.merge(credits,on='title')
```

```python
movies.head()
```

```python
# removinf unwanted columns. Jo chahiye wo ham yaha par rakh rahe hain
movies =  movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.head()
```

```python
# ab issse new data fram banayenge that has movie_id, title, tags(overview+genre+keywords+cast+crew))abs

movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
# firstly genre needs to be fixed
movies.iloc[0].genres# genre me we have strings so to convert it we use import ast eval
```

```python
import ast
```

```python
def convert(obj):
    l = []
    for i in ast.literal_eval(obj): # object ek list ban jaayega
        l.append(i['name'])
    return l
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
# same for  keywords column
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
# now for cast, we need first three names only
```

```python
def convert1(obj):
    l = []
    counter = 0
    for i in ast.literal_eval(obj): # object ek list ban jaayega
        if(counter!=3):
            l.append(i['name'])
            counter += 1
        else:
            break
    return l
```

```python
movies['cast'] = movies['cast'].apply(convert1)
```

```python
# for crew
def director(obj):
    l = []
    for i in ast.literal_eval(obj): 
        if(i['job'] == 'Director'):
            l.append(i['name'])
            break
    return l
```

```python
movies['crew'] = movies['crew'].apply(director)
```

```python
# converting strings of overview into list
movies['overview'] =  movies['overview'].apply(lambda x:x.split())
```

```python
# Now all the columns are in the list form, so we will concatenate them and then again at last change it into string
```

```python
# but hame sbke naam ke beech se pehe space htana pdega so that out model does not confuse
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","")for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","")for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","")for i in x][:3])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","")for i in x])
```

```python
movies['tag'] = movies['overview'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[['movie_id','title','tag']]
```

```python
new_df
```

```python
# now again converting to string
new_df['tag'] = new_df['tag'].apply(lambda x:" ".join(x))
```

```python
new_df
```

```python
# to lower case
new_df['tag'] = new_df['tag'].apply(lambda x:x.lower())
```

```python
# text vectorization
```

```python
# now we will use stemming
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
        return " ".join(y)
```

```python
new_df['tag'] = new_df['tag'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
# Fill any missing values in the 'tag' column with an empty string
new_df['tag'] = new_df['tag'].fillna("")
vectors = cv.fit_transform(new_df['tag']).toarray()
vectors
```

```python
# now we need to calculate distance between the vectors - jyada then utana hi different
# AND we will use cosine distance
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```
